## 9 Общие выводы 

Проведенная аналитика показывает, что датасет Natural Questions характеризуется несколькими важными свойствами.

Во-первых, в режиме первой аннотации преобладают примеры категории **NULL**, тогда как span-ответы составляют примерно треть корпуса.  
Во-вторых, наличие long answer распределено почти равномерно.  
В-третьих, сами short answers в большинстве случаев очень короткие, а значит основная сложность задачи связана не с генерацией длинного ответа,  
а с точной локализацией небольшого span внутри более крупного фрагмента.  
Наконец, распределения длины long answer показывают выраженный длинный хвост: большинство примеров компактны, но существует заметное число очень длинных фрагментов.
